```python
#in-built methods --> var_name.(after dot(.) long press 'Tab' button.
x='Hello World'
xcap=x.upper()
xsma=x.lower()
xtit=x.title()
print(xcap, xsma, xtit, sep='\t')
norspl=x.split()
ipspl=x.split('o')
print('normal split: ',norspl,' split with a specipic input:', ipspl)
```

```python
#string formatting
print('This is a string {}'.format('inserted'))
print('The {0} {0} {0}'.format('fox', 'brown', 'quick'))
print('The {q} {b} {f}'.format(f='fox', b='brown', q='quick'))
print('---------------------------------------------------------')

#another similar method
#f string literal method
name='Jose'
print(f'Hello, his name is {name}')
print('---------------------------------------------------------')
#number formatting
result=100/777
print("normal reusult is:",result)
print("the result was {r:1.3f}".format(r=result))

```

```python
# some other in-built methods (strings)
s='hello'
sco=s.count('l')#input is to be taken
sf=s.find('l')#only finds the first occurence of the input
sce=s.center(10,'z') #2 inputs , mark the result
print(sco, sf,sce)
s.expandtabs()#same as using sep='\t'


```

```python
#other string methods that return boolean values
q='variable'
t='1234'
pie='Ninja 123'
qa=q.isalpha()
ta=t.isalpha()
piea=pie.isalpha()
print(qa,ta,piea)
print('---------------------------------------------------------')
qspace=q.isspace()
pieend=pie.endswith('3')
piet=pie.istitle()
piel=pie.islower()
print(qspace, pieend, piet,piel)


```

```python
input2=int(input('number whose absolute value is to be checked:'))
abso=abs(input2)
input3=float(input('enter the number to be rounded up:'))
dec=int(input('to what decimal places would you like to round up the number other than the original one:'))
rou=round(input3)
rous=round(input3,dec)#to a specific number of places
print('absolute value:',abso)
print('rounded value as per your needs:')
print(rou, rous, sep='\t')
    
```

```python
#advanced sets
s=set()
s.add(1) #only one input at a time
s.add(2)
print(s)
print('----------------------------')
s.clear()
print(s)
print('----------------------------')
s1=set([1,2,3,4])
s2=set([3,4])
s3=set([4,5,6,7])
s2.add(2) 
s1.discard(1)
print(s1,s2)
print('----------------------------')
#other set operations
a=s2.issubset(s1)
b=s1.issuperset(s2)
c=s1.union(s3)
s2.update(s3)
print(f' s2 is a subset of s1: {a} s1 is a superset of s2: {b}')
print(f'the union of s1 and s3: {c} and the updated value of s2 is {s2}')
#remember there are some in-built methodds for set that if assigned to a varibale and if the variable is printed the output will be None eg; .update()
```
